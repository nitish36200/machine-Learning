```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv('data_file\insurance.csv')
df
```

```python
X = df[['age','bmi']]
y = df[['charges']]
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=20)
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
```

```python
model = lasso.fit(X_train,y_train)
```

```python
predicte = model.predict([[19,27.900]])
```

```python
print(predicte)
```

```python
sns.displot(predicte-y_test,kind='kde')
```
